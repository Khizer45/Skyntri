# Salary Prediction — Linear Regression

## Step 1 — Load the Dataset

```python
import pandas as pd

data = {
    'YearsOfExperience': [1, 2, 3, 4, 5],
    'Salary': [30, 35, 40, 45, 50]
}

df = pd.DataFrame(data)
print(df)
```

## Step 2 — Scatter Plot

```python
import matplotlib.pyplot as plt

plt.scatter(df['YearsOfExperience'], df['Salary'])
plt.xlabel('Years of Experience')
plt.ylabel('Salary (in 1000s)')
plt.title('Salary vs Years of Experience')
plt.show()
```

## Step 3 — Train the Linear Regression Model

```python
from sklearn.linear_model import LinearRegression

X = df[['YearsOfExperience']]
y = df['Salary']

model = LinearRegression()
model.fit(X, y)

print('Slope:', model.coef_[0])
print('Intercept:', model.intercept_)
```

## Step 4 — Predict Salary for 6 Years of Experience

```python
pred = model.predict([[6]])
print('Predicted Salary for 6 years of experience:', pred[0], 'thousand dollars')
```

## Step 5 — Plot Data Points and Regression Line

```python
import numpy as np

plt.scatter(df['YearsOfExperience'], df['Salary'], label='Actual Data')

x_line = np.linspace(0, 7, 100)
y_line = model.coef_[0] * x_line + model.intercept_
plt.plot(x_line, y_line, color='red', label='Regression Line')

plt.xlabel('Years of Experience')
plt.ylabel('Salary (in 1000s)')
plt.title('Salary Prediction — Linear Regression')
plt.legend()
plt.show()
```
